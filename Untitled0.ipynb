```python

from pathlib import Path
import pandas as pd

m = pd.read_csv("data/real/hursat/phailin_matched.csv")

m["time"] = pd.to_datetime(m["time"], utc=True)

def satellite_info(p):
    a = Path(p).name.split(".")
    return pd.Series({
        "satellite": a[7],
        "view_angle": float(a[6])
    })

m[["satellite", "view_angle"]] = (
    m["image_path"].apply(satellite_info)
)

print("Total matched files:", len(m))
print("Unique timestamps:", m["time"].nunique())

print("\nObservations by satellite:")
display(m.groupby("satellite").agg(
    files=("image_path", "count"),
    unique_times=("time", "nunique"),
    median_angle=("view_angle", "median")
))

display(m.head())
```

```python

import numpy as np
from netCDF4 import Dataset

def check_image(p):
    try:
        with Dataset(str(p)) as d:
            v = d.variables["IRWIN"]
            a = np.ma.squeeze(
                np.ma.asarray(v[:], dtype=np.float32)
            )

            a = a.filled(np.nan)

            valid = (
                np.isfinite(a) &
                (a >= 150) &
                (a <= 350)
            )

            return pd.Series({
                "shape_ok": a.shape == (301, 301),
                "valid_fraction": valid.mean(),
                "kelvin": str(v.units).lower() in
                          ["kelvin", "k"],
                "wv_available": "IRWVP" in d.variables
            })

    except Exception:
        return pd.Series({
            "shape_ok": False,
            "valid_fraction": 0.0,
            "kelvin": False,
            "wv_available": False
        })

q = m["image_path"].apply(check_image)
m = pd.concat([m, q], axis=1)

clean = m[
    m["shape_ok"] &
    m["kelvin"] &
    (m["valid_fraction"] == 1.0)
].copy()

print("Total files:", len(m))
print("Passed initial quality check:", len(clean))
print("Rejected files:", len(m) - len(clean))
print("Files containing WV variable:",
      clean["wv_available"].sum())

display(clean.head())
```

```python

assert not clean.empty, "No usable images found"

ranking = clean.groupby("satellite").agg(
    coverage=("time", "nunique"),
    angle=("view_angle", "median")
).sort_values(
    ["coverage", "angle"],
    ascending=[False, True]
)

sat = ranking.index[0]

u = clean[clean["satellite"] == sat].copy()

u = u.sort_values(["time", "view_angle"])
u = u.drop_duplicates(subset="time")
u = u.sort_values("time").reset_index(drop=True)

print("Selected satellite:", sat)
print("Unique usable observations:", len(u))
print("Remaining duplicate times:",
      u["time"].duplicated().sum())

u.to_csv(
    "data/real/hursat/phailin_clean.csv",
    index=False
)

display(u.head())
```

```python

import json

lookup = u.set_index("time")["image_path"]

seq = []

for _, r in labeled.iterrows():

    times = pd.date_range(
        end=r["time"],
        periods=9,
        freq="3h"
    )

    if not all(t in lookup.index for t in times):
        continue

    paths = [lookup.loc[t] for t in times]

    seq.append({
        "storm_id": r["storm_id"],
        "time": r["time"],
        "satellite": r["satellite"],
        "image_paths": json.dumps(paths),
        "wind_kt": r["wind_kt"],
        "future_wind_kt": r["future_wind_kt"],
        "ri": r["ri"],
        "kind": "real"
    })

sequences = pd.DataFrame(
    seq,
    columns=[
        "storm_id", "time", "satellite",
        "image_paths", "wind_kt",
        "future_wind_kt", "ri", "kind"
    ]
)

sequences.to_csv(
    "data/real/hursat/phailin_sequences.csv",
    index=False
)

print("Complete image sequences:", len(sequences))
print("RI sequences:", sequences["ri"].sum())
print("Independent cyclones:",
      sequences["storm_id"].nunique())

display(sequences.head())
```

```python

import numpy as np
import pandas as pd
from netCDF4 import Dataset

u = pd.read_csv("data/real/hursat/phailin_clean.csv")

results = []

for p in u["image_path"]:

    with Dataset(p) as d:

        r = {"image_path": p}

        for ch in ["IRWIN", "IRWVP"]:

            if ch not in d.variables:
                r[ch + "_valid"] = 0.0
                continue

            v = d.variables[ch]
            v.set_auto_maskandscale(True)

            a = np.ma.asarray(
                v[:], dtype=np.float32
            ).squeeze().filled(np.nan)

            good = (
                np.isfinite(a) &
                (a >= 150) &
                (a <= 350)
            )

            r[ch + "_valid"] = float(good.mean())
            r[ch + "_shape"] = str(a.shape)
            r[ch + "_units"] = str(
                getattr(v, "units", "")
            )

            r[ch + "_min"] = (
                float(a[good].min()) if good.any()
                else np.nan
            )

            r[ch + "_max"] = (
                float(a[good].max()) if good.any()
                else np.nan
            )

            r[ch + "_extra_calibration"] = str({
                k: str(v.getncattr(k))
                for k in v.ncattrs()
                if "cal" in k.lower()
            })

        results.append(r)

q = pd.DataFrame(results)

print("Total observations:", len(q))

print("Complete IR images:",
      (q["IRWIN_valid"] == 1).sum())

print("Complete WV images:",
      (q["IRWVP_valid"] == 1).sum())

print("\nCalibration attributes:")
print(q["IRWIN_extra_calibration"].value_counts())

display(q.drop(columns="image_path").head())

q.to_csv(
    "data/real/hursat/phailin_channel_quality.csv",
    index=False
)
```

```python

import json
import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
from netCDF4 import Dataset

seq = pd.read_csv(
    "data/real/hursat/phailin_sequences.csv"
)

def read_frame(p):

    channels = []

    with Dataset(p) as d:

        for ch in ["IRWIN", "IRWVP"]:

            v = d.variables[ch]
            v.set_auto_maskandscale(True)

            a = np.ma.asarray(
                v[:], dtype=np.float32
            ).squeeze().filled(np.nan)

            assert a.shape == (301, 301)
            assert np.isfinite(a).all()
            assert ((a >= 150) & (a <= 350)).all()

            channels.append(a)

    x = torch.from_numpy(np.stack(channels))

    x = F.interpolate(
        x.unsqueeze(0),
        size=(128, 128),
        mode="area"
    ).squeeze(0)

    return x.numpy()

paths = json.loads(seq.iloc[0]["image_paths"])

x = np.stack([read_frame(p) for p in paths])

print("Tensor shape:", x.shape)
print("Data type:", x.dtype)
print("Minimum temperature:", x.min())
print("Maximum temperature:", x.max())
print("All values finite:", np.isfinite(x).all())

assert x.shape == (9, 2, 128, 128)
```

```python

import requests
import pandas as pd
import re

from bs4 import BeautifulSoup
from urllib.parse import urljoin

tracks = pd.read_csv("data/real/tracks.csv")

ids = set(tracks["storm_id"].astype(str))

base = (
    "https://www.ncei.noaa.gov/data/"
    "hurricane-satellite-hursat-b1/archive/v06/"
)

found = []

for year in range(2013, 2016):

    url = base + str(year) + "/"

    try:
        r = requests.get(url, timeout=60)
        r.raise_for_status()

        soup = BeautifulSoup(r.text, "html.parser")

        for a in soup.find_all("a", href=True):

            href = a["href"]

            if not href.endswith(".tar.gz"):
                continue

            match = re.search(
                r"(\d{7}[NS]\d{5})", href
            )

            if not match:
                continue

            sid = match.group(1)

            if sid in ids:

                found.append({
                    "storm_id": sid,
                    "year": year,
                    "archive": href.split("/")[-1],
                    "url": urljoin(url, href)
                })

        print(year, "- Listing checked")

    except requests.RequestException as e:
        print(year, "- Error:", e)

catalogue = pd.DataFrame(
    found,
    columns=["storm_id", "year", "archive", "url"]
).drop_duplicates()

catalogue.to_csv(
    "data/real/hursat/available_storms.csv",
    index=False
)

print("\nAvailable matched cyclones:",
      catalogue["storm_id"].nunique())

display(catalogue)
```

```python

failed = []

for i, r in cat.iterrows():

    p = raw / r["archive"]
    tmp = p.with_suffix(".part")

    if p.exists() and tarfile.is_tarfile(p):
        print(i+1, "/", len(cat),
              "Already available:", r["storm_id"])
        continue

    try:
        with requests.get(
            r["url"], stream=True, timeout=(30, 120)
        ) as response:

            response.raise_for_status()

            with open(tmp, "wb") as f:
                for chunk in response.iter_content(
                    chunk_size=1024*1024
                ):
                    if chunk:
                        f.write(chunk)

        if not tarfile.is_tarfile(tmp):
            raise ValueError("Invalid archive")

        tmp.replace(p)

        print(i+1, "/", len(cat),
              "Downloaded:", r["storm_id"],
              round(p.stat().st_size / 1048576, 2),
              "MB")

    except Exception as e:
        failed.append(r["storm_id"])
        print("Failed:", r["storm_id"], str(e))

print("\nFailed downloads:", failed)
print("Available archives:", len(list(raw.glob("*.tar.gz"))))
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

b = Path("/content/drive/MyDrive/ChakraSense_Backup")

inventory = pd.read_csv(b / "all_observations.csv")

inventory["time"] = pd.to_datetime(
    inventory["time"], utc=True
)

good = inventory[
    inventory["ir_ok"].eq(True) &
    inventory["wv_ok"].eq(True)
].copy()

selected = []
summary = []

for sid, g in good.groupby("storm_id"):

    ranking = g.groupby("satellite").agg(
        coverage=("time", "nunique"),
        angle=("view_angle", "median")
    ).sort_values(
        ["coverage", "angle"],
        ascending=[False, True]
    )

    sat = ranking.index[0]

    x = g[g["satellite"] == sat].copy()

    x = x.sort_values(
        ["time", "view_angle", "member"]
    ).drop_duplicates("time")

    selected.append(x)

    summary.append({
        "storm_id": sid,
        "satellite": sat,
        "observations": len(x)
    })

selected = pd.concat(selected, ignore_index=True)

assert not selected.duplicated(
    ["storm_id", "time"]
).any()

selected.to_csv(
    b / "selected_observations.csv",
    index=False
)

display(pd.DataFrame(summary))

print("Selected observations:", len(selected))
print("Independent storms:",
      selected["storm_id"].nunique())

print("Duplicate storm timestamps:",
      selected.duplicated(["storm_id", "time"]).sum())
```

```python

import json

tracks = pd.read_csv(
    "/content/ChakraSense/ChakraSense/data/real/tracks.csv"
)

tracks["time"] = pd.to_datetime(
    tracks["time"], utc=True
)

tracks = tracks[
    (tracks["agency"] == "NEWDELHI") &
    (tracks["wind_averaging_minutes"] == 3)
].dropna(subset=["wind_kt"]).copy()

assert not tracks.duplicated(
    ["storm_id", "time"]
).any(), "Duplicate wind records"

rows = []

for sid, g in selected.groupby("storm_id"):

    g = g.sort_values("time")

    images = g.set_index("time")

    wind = tracks[
        tracks["storm_id"] == sid
    ].set_index("time")["wind_kt"]

    for t in images.index:

        times = pd.date_range(
            end=t,
            periods=9,
            freq="3h"
        )

        future = t + pd.Timedelta(hours=24)

        if not times.isin(images.index).all():
            continue

        if t not in wind.index or future not in wind.index:
            continue

        now = float(wind.loc[t])
        later = float(wind.loc[future])

        frames = images.loc[times]

        rows.append({
            "storm_id": sid,
            "year": int(g.iloc[0]["year"]),
            "time": t,
            "satellite": g.iloc[0]["satellite"],
            "archive": g.iloc[0]["archive"],
            "members": json.dumps(
                frames["member"].tolist()
            ),
            "wind_kt": now,
            "future_wind_kt": later,
            "wind_change_24h": later - now,
            "ri": int(later - now >= 30),
            "kind": "real"
        })

sequences = pd.DataFrame(rows)

assert not sequences.empty, "No complete sequences"

sequences.to_csv(
    b / "real_sequences_manifest.csv",
    index=False
)

print("Total complete sequences:", len(sequences))
print("Independent storms:",
      sequences["storm_id"].nunique())
print("RI sequences:", sequences["ri"].sum())
print("Non-RI sequences:",
      (sequences["ri"] == 0).sum())

display(sequences.head())
```

```python

import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
import tarfile
import json

from pathlib import Path
from netCDF4 import Dataset

b = Path("/content/drive/MyDrive/ChakraSense_Backup")
raw = b / "HURSAT_Archives"

out = b / "Real_Dataset_V2"
out.mkdir(parents=True, exist_ok=True)

seq = pd.read_csv(b / "real_sequences_with_splits.csv")

def convert_frame(content):

    channels = []

    with Dataset("memory.nc", memory=content) as d:

        for ch in ["IRWIN", "IRWVP"]:

            v = d.variables[ch]
            v.set_auto_maskandscale(True)

            a = np.ma.asarray(
                v[:], dtype=np.float32
            ).squeeze().filled(np.nan)

            assert a.shape == (301, 301)
            assert np.isfinite(a).all()
            assert ((a >= 150) & (a <= 350)).all()

            assert str(v.units).lower() in [
                "kelvin", "k"
            ]

            channels.append(a)

    x = torch.from_numpy(np.stack(channels))

    x = F.interpolate(
        x.unsqueeze(0),
        size=(128, 128),
        mode="area"
    ).squeeze(0)

    return x.numpy()


records = []

for sid, g in seq.groupby("storm_id"):

    assert g["archive"].nunique() == 1

    archive = raw / g.iloc[0]["archive"]

    needed = set()

    for text in g["members"]:
        needed.update(json.loads(text))

    frames = {}

    with tarfile.open(archive, "r|gz") as t:

        for m in t:

            if not m.isfile() or m.name not in needed:
                continue

            with t.extractfile(m) as f:
                frames[m.name] = convert_frame(f.read())

            if len(frames) == len(needed):
                break

    assert set(frames) == needed, (
        "Missing required frames: " + sid
    )

    names = sorted(frames)

    arrays = np.stack([
        frames[n] for n in names
    ]).astype(np.float32)

    index = {
        n: i for i, n in enumerate(names)
    }

    file = out / (sid + ".npz")

    np.savez_compressed(
        file,
        frames=arrays,
        members=np.array(names)
    )

    for _, r in g.iterrows():

        ids = [
            index[n]
            for n in json.loads(r["members"])
        ]

        assert len(ids) == 9

        records.append({
            "storm_id": sid,
            "time": r["time"],
            "satellite": r["satellite"],
            "kind": "real",
            "split": r["split"],
            "path": file.name,
            "frame_indices": json.dumps(ids),
            "wind_kt": r["wind_kt"],
            "future_wind_kt": r["future_wind_kt"],
            "ri": r["ri"]
        })

    print(
        sid,
        "| Frames:", len(arrays),
        "| Sequences:", len(g),
        "| Saved"
    )

manifest = pd.DataFrame(records)

manifest.to_csv(
    out / "samples.csv",
    index=False
)

assert len(manifest) == len(seq)

print("\nDataset preparation completed")
print("Total sequences:", len(manifest))
print("Independent storms:",
      manifest["storm_id"].nunique())

print("Saved to:", out)
```

```python

import requests
import tarfile

failed = []

for i, (_, r) in enumerate(cat.iterrows(), 1):

    p = raw_ext / r["archive"]
    tmp = raw_ext / (r["archive"] + ".part")

    # Skip previously downloaded archives
    if p.exists() and tarfile.is_tarfile(p):

        print(i, "/", len(cat),
              "Already available:", r["storm_id"])
        continue

    try:

        with requests.get(
            r["url"],
            stream=True,
            timeout=(30, 180)
        ) as response:

            response.raise_for_status()

            with open(tmp, "wb") as f:

                for chunk in response.iter_content(
                    chunk_size=1024*1024
                ):

                    if chunk:
                        f.write(chunk)

        if not tarfile.is_tarfile(tmp):
            raise ValueError("Invalid archive")

        tmp.replace(p)

        print(
            i, "/", len(cat),
            "Downloaded:", r["storm_id"],
            round(p.stat().st_size / 1048576, 2),
            "MB"
        )

    except Exception as e:

        failed.append(r["storm_id"])
        print("Failed:", r["storm_id"], str(e))

print("\nFailed downloads:", failed)

print("Available eligible archives:", sum(
    (raw_ext / n).exists() and
    tarfile.is_tarfile(raw_ext / n)
    for n in cat["archive"]
))

print("Total storage used:", round(
    sum(p.stat().st_size for p in raw_ext.glob("*.tar.gz"))
    / 1024**3, 2
), "GB")
```

```python

import numpy as np
import pandas as pd
import tarfile

from pathlib import Path
from netCDF4 import Dataset

b = Path("/content/drive/MyDrive/ChakraSense_Backup")
ext = b / "Historical_Expansion"
raw_ext = ext / "HURSAT_Archives"

cat = pd.read_csv(
    ext / "historical_download_catalogue.csv"
)

cat = cat[
    cat["labeled_records"].fillna(0) > 0
].drop_duplicates("archive")

# Save inspection results separately for each archive
qc = ext / "Quality_Checks"
qc.mkdir(parents=True, exist_ok=True)

columns = [
    "storm_id", "year", "time",
    "archive", "member", "satellite",
    "view_angle", "ir_ok", "wv_ok"
]

def check_channel(d, ch):

    if ch not in d.variables:
        return False

    v = d.variables[ch]
    v.set_auto_maskandscale(True)

    a = np.ma.asarray(
        v[:], dtype=np.float32
    ).squeeze().filled(np.nan)

    return bool(
        a.shape == (301, 301)
        and np.isfinite(a).all()
        and ((a >= 150) & (a <= 350)).all()
        and str(getattr(v, "units", "")).lower()
            in ["kelvin", "k"]
    )

for i, (_, r) in enumerate(cat.iterrows(), 1):

    sid = r["storm_id"]
    p = raw_ext / r["archive"]
    saved = qc / (r["archive"] + ".csv")

    if saved.exists():
        print(i, "/", len(cat), sid, "Already inspected")
        continue

    if not p.exists():
        print(sid, "Archive unavailable")
        continue

    rows = []
    errors = 0

    try:

        with tarfile.open(p, "r|gz") as t:

            for m in t:

                if not m.isfile() or not m.name.endswith(".nc"):
                    continue

                try:

                    a = Path(m.name).name.split(".")

                    if a[0] != sid:
                        continue

                    tm = pd.to_datetime(
                        "".join(a[2:6]),
                        format="%Y%m%d%H%M",
                        utc=True
                    )

                    with t.extractfile(m) as f:
                        content = f.read()

                    with Dataset(
                        "memory.nc", memory=content
                    ) as d:

                        ir = check_channel(d, "IRWIN")
                        wv = check_channel(d, "IRWVP")

                    rows.append({
                        "storm_id": sid,
                        "year": r["year"],
                        "time": tm,
                        "archive": r["archive"],
                        "member": m.name,
                        "satellite": a[7],
                        "view_angle": float(a[6]),
                        "ir_ok": ir,
                        "wv_ok": wv
                    })

                except Exception as e:
                    errors += 1
                    print("File error:", m.name, str(e))

        result = pd.DataFrame(rows, columns=columns)

        # Save only after finishing archive inspection
        tmp = saved.with_suffix(".part")
        result.to_csv(tmp, index=False)
        tmp.replace(saved)

        print(
            i, "/", len(cat), sid,
            "| Observations:", len(result),
            "| Errors:", errors
        )

    except Exception as e:

        print("Archive error:", sid, str(e))

print("\nInspection finished")
print("Inspection files:", len(list(qc.glob("*.csv"))))
```

```python

import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
import tarfile
import json

from pathlib import Path
from netCDF4 import Dataset

b = Path("/content/drive/MyDrive/ChakraSense_Backup")
ext = b / "Historical_Expansion"
raw = ext / "HURSAT_Archives"

out = ext / "Numerical_Dataset"
out.mkdir(parents=True, exist_ok=True)

seq = pd.read_csv(ext / "historical_sequences.csv")

def convert_frame(content):

    channels = []
    metadata = {}

    with Dataset("memory.nc", memory=content) as d:

        for ch in ["IRWIN", "IRWVP"]:

            v = d.variables[ch]
            v.set_auto_maskandscale(True)

            a = np.ma.asarray(
                v[:], dtype=np.float32
            ).squeeze().filled(np.nan)

            assert a.shape == (301, 301)
            assert np.isfinite(a).all()
            assert ((a >= 150) & (a <= 350)).all()

            assert str(v.units).lower() in [
                "kelvin", "k"
            ]

            metadata[ch] = {
                k: str(v.getncattr(k))
                for k in v.ncattrs()
            }

            channels.append(a)

    x = torch.from_numpy(np.stack(channels))

    x = F.interpolate(
        x.unsqueeze(0),
        size=(128, 128),
        mode="area"
    ).squeeze(0)

    return x.numpy(), json.dumps(metadata)


records = []
failed = []

for sid, g in seq.groupby("storm_id"):

    try:

        assert g["archive"].nunique() == 1

        archive = raw / g.iloc[0]["archive"]

        needed = set()

        for text in g["members"]:
            needed.update(json.loads(text))

        file = out / (sid + ".npz")
        names = None

        # Reuse successfully converted storm files
        if file.exists():

            try:

                with np.load(
                    file, allow_pickle=False
                ) as d:

                    saved_names = d["members"].tolist()
                    a = d["frames"]

                    if (
                        set(saved_names) == needed
                        and a.shape ==
                        (len(needed), 2, 128, 128)
                        and a.dtype == np.float32
                        and np.isfinite(a).all()
                        and "metadata" in d.files
                    ):
                        names = saved_names

            except Exception:
                names = None

        if names is None:

            frames = {}
            metadata = {}

            with tarfile.open(archive, "r|gz") as t:

                for m in t:

                    if not m.isfile():
                        continue

                    if m.name not in needed:
                        continue

                    with t.extractfile(m) as f:

                        x, info = convert_frame(
                            f.read()
                        )

                    frames[m.name] = x
                    metadata[m.name] = info

                    if len(frames) == len(needed):
                        break

            assert set(frames) == needed

            names = sorted(frames)

            arrays = np.stack([
                frames[n] for n in names
            ]).astype(np.float32)

            tmp = out / (sid + ".part.npz")

            np.savez_compressed(
                tmp,
                frames=arrays,
                members=np.array(names),
                metadata=np.array([
                    metadata[n] for n in names
                ])
            )

            tmp.replace(file)

        index = {
            n: i for i, n in enumerate(names)
        }

        for _, r in g.iterrows():

            ids = [
                index[n]
                for n in json.loads(r["members"])
            ]

            assert len(ids) == 9

            record = r.drop(
                labels=["members"]
            ).to_dict()

            record["path"] = file.name
            record["frame_indices"] = json.dumps(ids)

            records.append(record)

        print(
            sid,
            "| Frames:", len(names),
            "| Sequences:", len(g),
            "| Saved"
        )

    except Exception as e:

        failed.append({
            "storm_id": sid,
            "error": repr(e)
        })

        print("Failed:", sid, repr(e))

manifest_ext = pd.DataFrame(records)

assert not manifest_ext.empty

manifest_ext.to_csv(
    out / "samples.csv",
    index=False
)

pd.DataFrame(
    failed,
    columns=["storm_id", "error"]
).to_csv(
    out / "conversion_errors.csv",
    index=False
)

print("\nHistorical conversion completed")

print("Total numerical sequences:",
      len(manifest_ext))

print("Independent storms:",
      manifest_ext["storm_id"].nunique())

print("Failed storms:", len(failed))

print("Saved to:", out)
```

```python

class CycloneDataset(Dataset):

    def __init__(self, data, split):

        self.data = data[
            data["split"] == split
        ].reset_index(drop=True)

        self.cache = {}

    def __len__(self):

        return len(self.data)

    def __getitem__(self, i):

        r = self.data.iloc[i]

        path = r["file_path"]

        if path not in self.cache:

            with np.load(
                path, allow_pickle=False
            ) as d:

                self.cache[path] = d["frames"]

        ids = json.loads(r["frame_indices"])

        x = self.cache[path][ids].copy()

        # Fixed temperature normalization
        x = (x - 250.0) / 50.0

        return {
            "images": torch.tensor(
                x, dtype=torch.float32
            ),

            "current_wind": torch.tensor(
                r["wind_kt"],
                dtype=torch.float32
            ),

            "future_wind": torch.tensor(
                r["future_wind_kt"],
                dtype=torch.float32
            ),

            "ri": torch.tensor(
                r["ri"],
                dtype=torch.float32
            ),

            "storm_id": r["storm_id"]
        }


train_set = CycloneDataset(data, "train")
dev_set = CycloneDataset(data, "development")
test_set = CycloneDataset(data, "test")

print("Training samples:", len(train_set))
print("Development samples:", len(dev_set))
print("Testing samples:", len(test_set))

sample = train_set[0]

print("\nSample image shape:",
      sample["images"].shape)

print("Current wind:", sample["current_wind"])
print("Future wind:", sample["future_wind"])
print("RI label:", sample["ri"])

print("Finite input values:",
      torch.isfinite(sample["images"]).all().item())
```

```python

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import random

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed()

class ChakraSenseV2(nn.Module):

    def __init__(self):
        super().__init__()

        self.cnn = nn.Sequential(
            nn.Conv2d(2, 8, 5, stride=2, padding=2),
            nn.BatchNorm2d(8),
            nn.ReLU(),

            nn.Conv2d(8, 16, 3, stride=2, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),

            nn.Conv2d(16, 32, 3, stride=2, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),

            nn.Conv2d(32, 48, 3, stride=2, padding=1),
            nn.BatchNorm2d(48),
            nn.ReLU(),

            nn.AdaptiveAvgPool2d((2, 2)),
            nn.Flatten()
        )

        self.gru = nn.GRU(
            input_size=192,
            hidden_size=64,
            batch_first=True
        )

        self.fusion = nn.Sequential(
            nn.Linear(65, 64),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(64, 32),
            nn.ReLU()
        )

        self.wind_head = nn.Linear(32, 1)
        self.ri_head = nn.Linear(32, 1)

    def forward(self, images, current_wind):

        b, t, c, h, w = images.shape

        x = images.reshape(b*t, c, h, w)

        x = self.cnn(x)

        x = x.reshape(b, t, -1)

        _, hidden = self.gru(x)

        features = hidden[-1]

        wind = current_wind.reshape(-1, 1) / 100.0

        features = torch.cat(
            [features, wind], dim=1
        )

        features = self.fusion(features)

        delta = self.wind_head(features).squeeze(-1)
        ri_logits = self.ri_head(features).squeeze(-1)

        return delta, ri_logits


model = ChakraSenseV2().to(device)

print(model)

print("\nTrainable parameters:", sum(
    p.numel() for p in model.parameters()
    if p.requires_grad
))

print("Model device:", next(model.parameters()).device)
```

```python

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    average_precision_score,
    roc_auc_score,
    brier_score_loss
)

def train_epoch(model, loader, optimizer):

    model.train()

    total = 0
    count = 0

    for batch in loader:

        x = batch["images"].to(device)
        w = batch["current_wind"].to(device)
        y = batch["future_wind"].to(device)
        ri = batch["ri"].to(device)

        optimizer.zero_grad()

        delta, logits = model(x, w)

        target_delta = (y - w) / 30.0

        wind_loss = F.smooth_l1_loss(
            delta, target_delta
        )

        ri_loss = F.binary_cross_entropy_with_logits(
            logits, ri
        )

        loss = wind_loss + ri_loss

        loss.backward()

        nn.utils.clip_grad_norm_(
            model.parameters(), 1.0
        )

        optimizer.step()

        total += loss.item() * len(x)
        count += len(x)

    return total / count


@torch.no_grad()
def evaluate(model, loader):

    model.eval()

    actual = []
    predicted = []
    labels = []
    probabilities = []

    for batch in loader:

        x = batch["images"].to(device)
        w = batch["current_wind"].to(device)

        delta, logits = model(x, w)

        pred = torch.clamp(
            w + delta * 30.0, min=0
        )

        prob = torch.sigmoid(logits)

        actual.extend(
            batch["future_wind"].numpy()
        )

        predicted.extend(pred.cpu().numpy())

        labels.extend(batch["ri"].numpy())

        probabilities.extend(prob.cpu().numpy())

    return {
        "MAE": mean_absolute_error(
            actual, predicted
        ),
        "RMSE": np.sqrt(mean_squared_error(
            actual, predicted
        )),
        "AP": average_precision_score(
            labels, probabilities
        ),
        "ROC_AUC": roc_auc_score(
            labels, probabilities
        ),
        "Brier": brier_score_loss(
            labels, probabilities
        )
    }
```

```python

from pathlib import Path
import pandas as pd

save = Path(
    "/content/drive/MyDrive/"
    "ChakraSense_Backup/Real_CNN_GRU"
)

save.mkdir(parents=True, exist_ok=True)

set_seed(42)

model = ChakraSenseV2().to(device)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=0.0001
)

epochs = 25
patience = 6

best_score = float("inf")
best_epoch = 0
wait = 0
history = []

for epoch in range(1, epochs + 1):

    loss = train_epoch(
        model, fit_loader, optimizer
    )

    metrics = evaluate(model, val_loader)

    # Lower is better
    score = (
        metrics["MAE"] / 30.0
        + (1.0 - metrics["AP"])
    )

    history.append({
        "epoch": epoch,
        "train_loss": loss,
        "selection_score": score,
        **metrics
    })

    print(
        f"Epoch {epoch:02d} | "
        f"Loss: {loss:.4f} | "
        f"MAE: {metrics['MAE']:.2f} | "
        f"RMSE: {metrics['RMSE']:.2f} | "
        f"AP: {metrics['AP']:.4f} | "
        f"Brier: {metrics['Brier']:.4f}"
    )

    pd.DataFrame(history).to_csv(
        save / "training_history.csv",
        index=False
    )

    if score < best_score:

        best_score = score
        best_epoch = epoch
        wait = 0

        torch.save(
            model.state_dict(),
            save / "internal_best_model.pt"
        )

        (save / "best_epoch.txt").write_text(
            str(best_epoch)
        )

        print("Best internal model saved")

    else:

        wait += 1

    if wait >= patience:

        print("Early stopping activated")
        break

print("\nBest epoch:", best_epoch)
print("Best selection score:", best_score)
print("Internal training completed")
```

```python

import numpy as np
import pandas as pd

def extract_features(ds):

    table = ds.data.copy()

    times = pd.to_datetime(
        table["time"], utc=True
    )

    lookup = dict(zip(
        zip(table["storm_id"], times),
        table["wind_kt"]
    ))

    rows = []

    for i in range(len(ds)):

        r = table.iloc[i]
        sample = ds[i]

        # Convert normalized temperatures back to Kelvin
        x = sample["images"].numpy() * 50 + 250

        now = float(r["wind_kt"])
        t = times.iloc[i]

        f = {
            "current_wind": now
        }

        # Previous historical wind changes
        for h in [6, 12, 24]:

            previous = lookup.get(
                (
                    r["storm_id"],
                    t - pd.Timedelta(hours=h)
                ),
                np.nan
            )

            f[f"wind_change_{h}h"] = now - previous

        # Extract features from both channels
        for j, ch in enumerate(["IR", "WV"]):

            first = x[0, j]
            last = x[-1, j]

            centre = last[40:88, 40:88]

            f[f"{ch}_mean"] = last.mean()
            f[f"{ch}_std"] = last.std()
            f[f"{ch}_min"] = last.min()

            f[f"{ch}_centre_mean"] = centre.mean()

            f[f"{ch}_change"] = (
                last.mean() - first.mean()
            )

            f[f"{ch}_p10"] = np.percentile(last, 10)

        f["IR_cold_cloud_fraction"] = (
            x[-1, 0] < 230
        ).mean()

        rows.append(f)

    return pd.DataFrame(rows)


X_train_new = extract_features(train_set)
X_dev_new = extract_features(dev_set)

print("Training feature shape:", X_train_new.shape)
print("Development feature shape:", X_dev_new.shape)

print("\nExtracted features:")
print(X_train_new.columns.tolist())

print("\nMissing feature values:")
print(X_train_new.isna().sum())

X_train_new.to_csv(
    opt_dir / "training_features.csv",
    index=False
)

X_dev_new.to_csv(
    opt_dir / "development_features.csv",
    index=False
)

display(X_train_new.head())
```

```python

from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, LogisticRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    average_precision_score,
    roc_auc_score,
    brier_score_loss
)

import joblib

y_train = train_set.data["future_wind_kt"].to_numpy()
y_dev = dev_set.data["future_wind_kt"].to_numpy()

ri_train = train_set.data["ri"].to_numpy()
ri_dev = dev_set.data["ri"].to_numpy()

w_train = train_set.data["wind_kt"].to_numpy()
w_dev = dev_set.data["wind_kt"].to_numpy()

wind_cols = [
    "current_wind",
    "wind_change_6h",
    "wind_change_12h",
    "wind_change_24h"
]

feature_groups = {
    "Wind_History": wind_cols,
    "Wind_and_Images": X_train_new.columns.tolist()
}

results = []

for name, cols in feature_groups.items():

    ridge = make_pipeline(
        SimpleImputer(
            strategy="median",
            add_indicator=True
        ),
        StandardScaler(),
        Ridge(alpha=10)
    )

    logistic = make_pipeline(
        SimpleImputer(
            strategy="median",
            add_indicator=True
        ),
        StandardScaler(),
        LogisticRegression(
            C=0.1,
            max_iter=2000
        )
    )

    ridge.fit(
        X_train_new[cols],
        y_train - w_train
    )

    logistic.fit(
        X_train_new[cols],
        ri_train
    )

    wind_pred = np.maximum(
        w_dev + ridge.predict(X_dev_new[cols]),
        0
    )

    ri_prob = logistic.predict_proba(
        X_dev_new[cols]
    )[:, 1]

    results.append({
        "Model": name,
        "MAE": mean_absolute_error(
            y_dev, wind_pred
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_dev, wind_pred)
        ),
        "AP": average_precision_score(
            ri_dev, ri_prob
        ),
        "ROC_AUC": roc_auc_score(
            ri_dev, ri_prob
        ),
        "Brier": brier_score_loss(
            ri_dev, ri_prob
        )
    })

    joblib.dump(
        {
            "wind_model": ridge,
            "ri_model": logistic,
            "features": cols
        },
        opt_dir / f"{name}.joblib"
    )

new_results = pd.DataFrame(results)

old_results = pd.read_csv(
    save / "development_model_comparison.csv"
)

comparison = pd.concat(
    [old_results, new_results],
    ignore_index=True
)

display(
    comparison.set_index("Model").round(4)
)

comparison.to_csv(
    opt_dir / "improved_model_comparison.csv",
    index=False
)

print("Improved baseline models saved")
```

```python

import pandas as pd
import numpy as np

from pathlib import Path
from sklearn.model_selection import StratifiedKFold

b = Path("/content/drive/MyDrive/ChakraSense_Backup")

dest = b / "Combined_Real_Dataset"
opt_dir = b / "Model_Optimization"

cv_dir = b / "Cross_Validation"
cv_dir.mkdir(parents=True, exist_ok=True)

data = pd.read_csv(
    dest / "combined_samples_with_splits.csv"
)

train = data[
    data["split"] == "train"
].reset_index(drop=True)

X = pd.read_csv(
    opt_dir / "training_features.csv"
)

assert len(X) == len(train)

assert np.allclose(
    X["current_wind"], train["wind_kt"]
)

storm_table = train.groupby("storm_id").agg(
    sequences=("ri", "size"),
    has_ri=("ri", "max")
).reset_index()

skf = StratifiedKFold(
    n_splits=4,
    shuffle=True,
    random_state=42
)

folds = []

train["fold"] = -1

for fold, (a, v) in enumerate(
    skf.split(storm_table, storm_table["has_ri"]),
    start=1
):

    fit_ids = storm_table.iloc[a]["storm_id"]
    val_ids = storm_table.iloc[v]["storm_id"]

    fit_idx = np.flatnonzero(
        train["storm_id"].isin(fit_ids)
    )

    val_idx = np.flatnonzero(
        train["storm_id"].isin(val_ids)
    )

    assert not set(fit_ids) & set(val_ids)

    train.loc[val_idx, "fold"] = fold

    folds.append((fit_idx, val_idx))

    print(f"\nFOLD {fold}")
    print("Training storms:", len(fit_ids))
    print("Validation storms:", len(val_ids))

    print("Training sequences:", len(fit_idx))
    print("Validation sequences:", len(val_idx))

    print("Validation RI sequences:",
          int(train.iloc[val_idx]["ri"].sum()))

    print("Storm overlap: None")

assert (train["fold"] > 0).all()

train.to_csv(
    cv_dir / "training_fold_assignments.csv",
    index=False
)

print("\nTotal folds:", len(folds))
print("Cross-validation groups saved successfully")
```

```python

from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, LogisticRegression

from sklearn.ensemble import (
    HistGradientBoostingRegressor,
    HistGradientBoostingClassifier
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    average_precision_score,
    roc_auc_score,
    brier_score_loss
)

wind_cols = [
    "current_wind",
    "wind_change_6h",
    "wind_change_12h",
    "wind_change_24h"
]

image_cols = [
    c for c in X.columns
    if c.startswith(("IR_", "WV_"))
]

configs = {
    "Current_Wind": ["current_wind"],
    "Wind_History": wind_cols,
    "Current_Wind_Images": [
        "current_wind"
    ] + image_cols,
    "Wind_and_Images": X.columns.tolist(),
    "Gradient_Boosting": X.columns.tolist()
}

y = train["future_wind_kt"].to_numpy()
w = train["wind_kt"].to_numpy()
ri = train["ri"].to_numpy()

results = []
predictions = []

def record(name, fold, v, wp, rp):

    results.append({
        "Model": name,
        "Fold": fold,
        "MAE": mean_absolute_error(y[v], wp),
        "RMSE": np.sqrt(
            mean_squared_error(y[v], wp)
        ),
        "AP": average_precision_score(
            ri[v], rp
        ),
        "ROC_AUC": roc_auc_score(
            ri[v], rp
        ),
        "Brier": brier_score_loss(
            ri[v], rp
        )
    })

    p = train.iloc[v][[
        "storm_id", "time", "wind_kt",
        "future_wind_kt", "ri"
    ]].copy()

    p["Model"] = name
    p["Fold"] = fold
    p["predicted_wind"] = wp
    p["ri_probability"] = rp

    predictions.append(p)


for fold, (a, v) in enumerate(folds, start=1):

    print(f"\nProcessing fold {fold}")

    record(
        "Persistence_Climatology",
        fold, v, w[v],
        np.full(len(v), ri[a].mean())
    )

    for name, cols in configs.items():

        if name == "Gradient_Boosting":

            reg = HistGradientBoostingRegressor(
                learning_rate=0.05,
                max_iter=100,
                max_leaf_nodes=7,
                min_samples_leaf=30,
                l2_regularization=5,
                early_stopping=False,
                random_state=42
            )

            clf = HistGradientBoostingClassifier(
                learning_rate=0.05,
                max_iter=100,
                max_leaf_nodes=7,
                min_samples_leaf=30,
                l2_regularization=5,
                early_stopping=False,
                random_state=42
            )

        else:

            reg = make_pipeline(
                SimpleImputer(
                    strategy="median",
                    add_indicator=True
                ),
                StandardScaler(),
                Ridge(alpha=10)
            )

            clf = make_pipeline(
                SimpleImputer(
                    strategy="median",
                    add_indicator=True
                ),
                StandardScaler(),
                LogisticRegression(
                    C=0.1,
                    max_iter=2000
                )
            )

        reg.fit(
            X.iloc[a][cols],
            y[a] - w[a]
        )

        clf.fit(
            X.iloc[a][cols],
            ri[a]
        )

        wp = np.maximum(
            w[v] + reg.predict(X.iloc[v][cols]),
            0
        )

        rp = clf.predict_proba(
            X.iloc[v][cols]
        )[:, 1]

        record(name, fold, v, wp, rp)

        print(name, "completed")

cv_results = pd.DataFrame(results)

cv_results.to_csv(
    cv_dir / "fold_results.csv",
    index=False
)

pd.concat(
    predictions,
    ignore_index=True
).to_csv(
    cv_dir / "out_of_fold_predictions.csv",
    index=False
)

print("\nCross-validation completed successfully")

display(cv_results.round(4))
```

```python

data = pd.read_csv(
    dest / "combined_samples_with_splits.csv"
)

train = data[
    data["split"] == "train"
].reset_index(drop=True)

dev = data[
    data["split"] == "development"
].reset_index(drop=True)

X_train_old = pd.read_csv(
    opt_dir / "training_features.csv"
)

X_dev_old = pd.read_csv(
    opt_dir / "development_features.csv"
)

assert len(train) == len(X_train_old)
assert len(dev) == len(X_dev_old)

assert np.allclose(
    train["wind_kt"],
    X_train_old["current_wind"]
)

assert np.allclose(
    dev["wind_kt"],
    X_dev_old["current_wind"]
)

lookup = {}

for sid, g in original.groupby("storm_id"):

    g = g.sort_values("time")

    g = g.drop_duplicates(
        "time", keep="last"
    )

    lookup[sid] = (
        pd.DatetimeIndex(g["time"]),
        g["wind_kt"].to_numpy()
    )


def previous_wind(sid, t):

    if sid not in lookup:
        return np.nan, np.nan

    times, winds = lookup[sid]

    i = times.searchsorted(
        t, side="right"
    ) - 1

    if i < 0:
        return np.nan, np.nan

    age = (
        t - times[i]
    ).total_seconds() / 3600

    if age > 6:
        return np.nan, np.nan

    return float(winds[i]), age


def improve_features(table, old):

    X = old.copy()

    rows = []

    for _, r in table.iterrows():

        sid = r["storm_id"]
        t = pd.to_datetime(r["time"], utc=True)

        now, age = previous_wind(sid, t)

        f = {
            "current_wind": now,
            "current_wind_age": age
        }

        for h in [6, 12, 24]:

            past, past_age = previous_wind(
                sid,
                t - pd.Timedelta(hours=h)
            )

            f[f"wind_change_{h}h"] = now - past
            f[f"wind_age_{h}h"] = past_age

        rows.append(f)

    improved = pd.DataFrame(rows)

    for c in improved.columns:
        X[c] = improved[c]

    return X


X_train_v3 = improve_features(
    train, X_train_old
)

X_dev_v3 = improve_features(
    dev, X_dev_old
)

print("Training feature shape:",
      X_train_v3.shape)

print("Development feature shape:",
      X_dev_v3.shape)

print("\nPrevious missing wind trends:")

cols = [
    "wind_change_6h",
    "wind_change_12h",
    "wind_change_24h"
]

print(X_train_old[cols].isna().sum())

print("\nNew missing wind trends:")
print(X_train_v3[cols].isna().sum())

print("\nAll missing feature values:")
print(X_train_v3.isna().sum())

X_train_v3.to_csv(
    v3_dir / "training_features_v3.csv",
    index=False
)

X_dev_v3.to_csv(
    v3_dir / "development_features_v3.csv",
    index=False
)

print("\nV3 features saved successfully")

display(X_train_v3.head())
```

```python

from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, LogisticRegression
from sklearn.ensemble import (
    HistGradientBoostingRegressor,
    HistGradientBoostingClassifier
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    average_precision_score,
    roc_auc_score,
    brier_score_loss
)

y = train["future_wind_kt"].to_numpy()
w = train["wind_kt"].to_numpy()
ri = train["ri"].to_numpy()

wind_cols = [
    "current_wind",
    "wind_change_6h",
    "wind_change_12h",
    "wind_change_24h",
    "current_wind_age",
    "wind_age_6h",
    "wind_age_12h",
    "wind_age_24h"
]

image_cols = [
    c for c in X.columns
    if c.startswith(("IR_", "WV_"))
]

configs = {
    "Current_Wind": ["current_wind"],
    "Wind_History": wind_cols,
    "Current_Wind_Images": [
        "current_wind"
    ] + image_cols,
    "Wind_and_Images": X.columns.tolist(),
    "Gradient_Boosting": X.columns.tolist()
}

results = []

for fold in sorted(fold_data["fold"].unique()):

    a = np.flatnonzero(
        fold_data["fold"].to_numpy() != fold
    )

    v = np.flatnonzero(
        fold_data["fold"].to_numpy() == fold
    )

    assert not (
        set(train.iloc[a]["storm_id"])
        & set(train.iloc[v]["storm_id"])
    )

    print(f"\nProcessing V3 fold {fold}")

    # Persistence + climatology benchmark
    candidates = [(
        "Persistence_Climatology",
        w[v],
        np.full(len(v), ri[a].mean())
    )]

    for name, cols in configs.items():

        if name == "Gradient_Boosting":

            reg = HistGradientBoostingRegressor(
                learning_rate=0.05,
                max_iter=100,
                max_leaf_nodes=7,
                min_samples_leaf=30,
                l2_regularization=5,
                early_stopping=False,
                random_state=42
            )

            clf = HistGradientBoostingClassifier(
                learning_rate=0.05,
                max_iter=100,
                max_leaf_nodes=7,
                min_samples_leaf=30,
                l2_regularization=5,
                early_stopping=False,
                random_state=42
            )

        else:

            reg = make_pipeline(
                SimpleImputer(
                    strategy="median",
                    add_indicator=True
                ),
                StandardScaler(),
                Ridge(alpha=10)
            )

            clf = make_pipeline(
                SimpleImputer(
                    strategy="median",
                    add_indicator=True
                ),
                StandardScaler(),
                LogisticRegression(
                    C=0.1,
                    max_iter=2000
                )
            )

        reg.fit(
            X.iloc[a][cols],
            y[a] - w[a]
        )

        clf.fit(
            X.iloc[a][cols],
            ri[a]
        )

        wp = np.maximum(
            w[v] + reg.predict(X.iloc[v][cols]),
            0
        )

        rp = clf.predict_proba(
            X.iloc[v][cols]
        )[:, 1]

        candidates.append((name, wp, rp))

    for name, wp, rp in candidates:

        results.append({
            "Model": name,
            "Fold": fold,
            "MAE": mean_absolute_error(y[v], wp),
            "RMSE": np.sqrt(
                mean_squared_error(y[v], wp)
            ),
            "AP": average_precision_score(ri[v], rp),
            "ROC_AUC": roc_auc_score(ri[v], rp),
            "Brier": brier_score_loss(ri[v], rp)
        })

    print("Fold completed")

v3_results = pd.DataFrame(results)

v3_results.to_csv(
    v3 / "v3_cross_validation_results.csv",
    index=False
)

print("\nSTEP 74B MODEL TRAINING COMPLETED")
display(v3_results.round(4))
```

```python

metrics = [
    "MAE", "RMSE", "AP", "ROC_AUC", "Brier"
]

old = pd.read_csv(
    cv / "fold_results.csv"
)

old_avg = old.groupby("Model")[metrics].mean()

new_avg = v3_results.groupby("Model")[metrics].mean()

comparison = old_avg.join(
    new_avg,
    lsuffix="_old",
    rsuffix="_v3"
)

comparison["MAE_improvement"] = (
    comparison["MAE_old"] -
    comparison["MAE_v3"]
)

comparison["RMSE_improvement"] = (
    comparison["RMSE_old"] -
    comparison["RMSE_v3"]
)

comparison["AP_gain"] = (
    comparison["AP_v3"] -
    comparison["AP_old"]
)

comparison["ROC_AUC_gain"] = (
    comparison["ROC_AUC_v3"] -
    comparison["ROC_AUC_old"]
)

comparison["Brier_improvement"] = (
    comparison["Brier_old"] -
    comparison["Brier_v3"]
)

print("STEP 72 VS STEP 74B COMPARISON")
display(comparison.round(4))

print("\nV3 AVERAGE PERFORMANCE")
display(new_avg.round(4))

print("\nV3 FOLD-TO-FOLD VARIATION")
display(
    v3_results.groupby("Model")[metrics].std().round(4)
)

comparison.to_csv(
    v3 / "v2_vs_v3_comparison.csv"
)

print("\nSTEP 74B COMPLETED SUCCESSFULLY")
```

```python

import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss
)

# Restore previous results if variables already exist
b = Path("/content/drive/MyDrive/ChakraSense_Backup")
v3 = b / "ChakraSense_V3"

assert len(X) == len(train) == len(fold_data)

# Ensure original current wind is used, as in Step 74B
X = X.copy()
X["current_wind"] = train["wind_kt"].to_numpy()

labels = train["ri"].astype(int).to_numpy()
folds = fold_data["fold"].to_numpy()

# Candidate model settings
configs = {
    "Original": dict(
        learning_rate=0.05,
        max_iter=100,
        max_leaf_nodes=7,
        min_samples_leaf=30,
        l2_regularization=5
    ),
    "Regularized": dict(
        learning_rate=0.05,
        max_iter=60,
        max_leaf_nodes=5,
        min_samples_leaf=40,
        l2_regularization=10
    ),
    "Slow_Learning": dict(
        learning_rate=0.035,
        max_iter=140,
        max_leaf_nodes=5,
        min_samples_leaf=25,
        l2_regularization=5
    ),
    "Balanced": dict(
        learning_rate=0.05,
        max_iter=100,
        max_leaf_nodes=7,
        min_samples_leaf=30,
        l2_regularization=5
    )
}

predictions = pd.DataFrame({
    "storm_id": train["storm_id"].to_numpy(),
    "actual_ri": labels,
    "fold": folds
})

fold_results = []

for name, params in configs.items():

    print("\nTraining:", name)

    probability = np.full(
        len(labels), np.nan
    )

    for f in sorted(np.unique(folds)):

        a = np.flatnonzero(folds != f)
        val = np.flatnonzero(folds == f)

        assert not (
            set(train.iloc[a]["storm_id"])
            & set(train.iloc[val]["storm_id"])
        )

        model = HistGradientBoostingClassifier(
            **params,
            early_stopping=False,
            random_state=42
        )

        if name == "Balanced":
            weights = compute_sample_weight(
                class_weight="balanced",
                y=labels[a]
            )
            model.fit(
                X.iloc[a],
                labels[a],
                sample_weight=weights
            )
        else:
            model.fit(
                X.iloc[a],
                labels[a]
            )

        p = model.predict_proba(
            X.iloc[val]
        )[:, 1]

        probability[val] = p

        fold_results.append({
            "Model": name,
            "Fold": int(f),
            "AP": average_precision_score(
                labels[val], p
            ),
            "ROC_AUC": roc_auc_score(
                labels[val], p
            ),
            "Brier": brier_score_loss(
                labels[val], p
            )
        })

        print(" Fold", f, "completed")

    assert np.isfinite(probability).all()

    predictions[name] = probability

# Evaluate pooled out-of-fold predictions
summary = []

for name in configs:

    p = predictions[name].to_numpy()

    summary.append({
        "Model": name,
        "AP": average_precision_score(labels, p),
        "ROC_AUC": roc_auc_score(labels, p),
        "Brier": brier_score_loss(labels, p)
    })

summary = pd.DataFrame(summary)
summary = summary.sort_values(
    "AP", ascending=False
)

display(summary.round(4))

# Save for later analysis
predictions.to_csv(
    v3 / "v3_ri_oof_predictions.csv",
    index=False
)

pd.DataFrame(fold_results).to_csv(
    v3 / "v3_ri_optimization_folds.csv",
    index=False
)

summary.to_csv(
    v3 / "v3_ri_optimization_summary.csv",
    index=False
)

print("\nSTEP 74C.1 COMPLETED SUCCESSFULLY")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

v3 = Path(
    "/content/drive/MyDrive/"
    "ChakraSense_Backup/ChakraSense_V3"
)

p = pd.read_csv(
    v3 / "v3_ri_oof_predictions.csv"
)

models = ["Regularized", "Slow_Learning"]
thresholds = [0.10, 0.15, 0.20, 0.25, 0.30, 0.40, 0.50]

results = []

y = p["actual_ri"].astype(int)

for model in models:
    probabilities = p[model].to_numpy()

    for t in thresholds:
        pred = (probabilities >= t).astype(int)

        tp = int(((pred == 1) & (y == 1)).sum())
        fp = int(((pred == 1) & (y == 0)).sum())
        fn = int(((pred == 0) & (y == 1)).sum())

        results.append({
            "Model": model,
            "Threshold": t,
            "Precision": precision_score(
                y, pred, zero_division=0
            ),
            "Recall": recall_score(
                y, pred, zero_division=0
            ),
            "F1": f1_score(
                y, pred, zero_division=0
            ),
            "True_Alerts": tp,
            "False_Alarms": fp,
            "Missed_RI": fn
        })

r = pd.DataFrame(results)

print("RI THRESHOLD COMPARISON")
display(r.round(4))

r.to_csv(
    v3 / "v3_threshold_comparison.csv",
    index=False
)

print("\nBEST EXPLORATORY F1 THRESHOLD")

best = r.loc[r["F1"].idxmax()]

display(best.to_frame().T)

print("\nSTEP 74E COMPLETED SUCCESSFULLY")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss
)

v3 = Path(
    "/content/drive/MyDrive/ChakraSense_Backup/ChakraSense_V3"
)

p = pd.read_csv(v3 / "v3_ri_oof_predictions.csv")

y = p["actual_ri"].astype(int)
prob = p["Regularized"].astype(float)

threshold = 0.15

p["pred"] = (prob >= threshold).astype(int)
p["fp"] = ((p["pred"] == 1) & (y == 0)).astype(int)
p["fn"] = ((p["pred"] == 0) & (y == 1)).astype(int)
p["tp"] = ((p["pred"] == 1) & (y == 1)).astype(int)

print("STEP 74G - MODEL DIAGNOSTICS")

print("\nFOLD-WISE PERFORMANCE")

rows = []

for fold, g in p.groupby("fold"):
    a = g["actual_ri"].astype(int)
    b = g["Regularized"].astype(float)

    rows.append({
        "fold": fold,
        "samples": len(g),
        "ri_cases": int(a.sum()),
        "false_positives": int(g["fp"].sum()),
        "missed_ri": int(g["fn"].sum()),
        "AP": average_precision_score(a, b),
        "ROC_AUC": (
            roc_auc_score(a, b)
            if a.nunique() == 2 else np.nan
        ),
        "Brier": brier_score_loss(a, b)
    })

fold_report = pd.DataFrame(rows)
display(fold_report)

print("\nPROBABILITY CALIBRATION")

p["probability_bin"] = pd.cut(
    prob,
    bins=[0, 0.1, 0.15, 0.2, 0.3, 0.4, 0.5, 1.000001],
    include_lowest=True,
    right=False
)

calibration = p.groupby(
    "probability_bin", observed=True
).agg(
    samples=("actual_ri", "size"),
    observed_ri_rate=("actual_ri", "mean"),
    mean_probability=("Regularized", "mean"),
    false_positives=("fp", "sum"),
    true_positives=("tp", "sum")
)

display(calibration)

print("\nOVERALL PROBABILITY CHECK")
print("Actual RI frequency:", round(y.mean(), 4))
print("Mean predicted probability:", round(prob.mean(), 4))
print("Brier score:", round(brier_score_loss(y, prob), 4))

fold_report.to_csv(
    v3 / "v3_fold_diagnostics.csv",
    index=False
)

calibration.to_csv(
    v3 / "v3_probability_calibration.csv"
)

print("\nSTEP 74G COMPLETED")
```

```python

# STEP 75E - FLEXIBLE TEMPORAL FILTER
# No GPU required

import numpy as np
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score

print("STEP 75E - FLEXIBLE TEMPORAL FILTER")
print("=" * 55)

# Restore predictions from Step 75B
cols = ["storm_id", "time", "actual_ri", "Regularized"]
assert all(c in pred.columns for c in cols)

df = pred[cols].copy()
df["time"] = pd.to_datetime(df["time"], utc=True)

assert df[cols].notna().all().all()
assert df["actual_ri"].isin([0, 1]).all()
assert df["Regularized"].between(0, 1).all()

df = df.sort_values(
    ["storm_id", "time"]
).reset_index(drop=True)

assert not df.duplicated(["storm_id", "time"]).any()

# Create independent continuous 3-hour segments
new_run = (
    df["storm_id"].ne(df["storm_id"].shift()) |
    df["time"].diff().ne(pd.Timedelta(hours=3))
)
df["run"] = new_run.cumsum()

prob = df["Regularized"]
actual = df["actual_ri"].astype(int)

# Original alert rule
original = prob >= 0.15

# Strict filter from Step 75D
strict_count = original.groupby(df["run"]).transform(
    lambda s: s.rolling(3, min_periods=2).sum()
)
strict = original & strict_count.ge(2)

# Highest probability in previous two observations
# Future predictions are never used
prior_max = prob.groupby(df["run"]).transform(
    lambda s: s.shift(1).rolling(
        window=2, min_periods=1
    ).max()
)

# True RI storm = at least one RI-positive window
storm_truth = actual.groupby(df["storm_id"]).max()

def evaluate(name, alert):
    alert = pd.Series(
        alert, index=df.index
    ).fillna(False).astype(bool)

    tp = int((alert & actual.eq(1)).sum())
    fp = int((alert & actual.eq(0)).sum())
    fn = int((~alert & actual.eq(1)).sum())

    # A storm is correctly detected only if an alert
    # coincides with an actual RI-positive window.
    hits = alert & actual.eq(1)
    storm_hits = hits.groupby(df["storm_id"]).any()

    all_alerts = alert.groupby(df["storm_id"]).any()

    detected = int(storm_hits[storm_truth.eq(1)].sum())
    false_storms = int(
        all_alerts[storm_truth.eq(0)].sum()
    )

    return {
        "Method": name,
        "Precision": precision_score(
            actual, alert, zero_division=0
        ),
        "Recall": recall_score(
            actual, alert, zero_division=0
        ),
        "F1": f1_score(
            actual, alert, zero_division=0
        ),
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "RI_Storms_Detected": detected,
        "RI_Storms_Missed": int(
            storm_truth.sum() - detected
        ),
        "Non_RI_Storms_Alerted": false_storms
    }

rows = [
    evaluate("Original", original),
    evaluate("Strict_75D", strict)
]

# Test different minimum previous probabilities
for weak in [0.05, 0.08, 0.10, 0.12, 0.15]:
    alert = original & prior_max.ge(weak)

    rows.append(
        evaluate(f"Support_{weak:.2f}", alert)
    )

results_75e = pd.DataFrame(rows)

print("\nWINDOW AND STORM-LEVEL RESULTS")
display(results_75e.round(4))

print("\nFALSE ALARM REDUCTION")
base_fp = results_75e.iloc[0]["FP"]

results_75e["FP_Reduction_%"] = (
    100 * (base_fp - results_75e["FP"]) / base_fp
)

display(
    results_75e[
        ["Method", "FP_Reduction_%", "Recall",
         "F1", "RI_Storms_Detected",
         "RI_Storms_Missed"]
    ].round(3)
)

print("\nSTEP 75E COMPLETED")
```

```python

# STEP 75F - MODEL ENSEMBLE EXPERIMENT

import numpy as np
import pandas as pd
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    average_precision_score, roc_auc_score,
    brier_score_loss
)

print("STEP 75F - MODEL ENSEMBLE EXPERIMENT")
print("=" * 55)

needed = [
    "storm_id", "actual_ri",
    "Regularized", "Slow_Learning"
]

assert all(c in pred.columns for c in needed)

d = pred[needed].copy().reset_index(drop=True)

assert len(d) == 1265
assert d.notna().all().all()
assert d["actual_ri"].isin([0, 1]).all().all()

y = d["actual_ri"].astype(int)

r = d["Regularized"].astype(float)
s = d["Slow_Learning"].astype(float)

assert r.between(0, 1).all()
assert s.between(0, 1).all()

models = {
    "Regularized": r,
    "Slow_Learning": s,
    "Ensemble_50_50": 0.5*r + 0.5*s,
    "Ensemble_70_30": 0.7*r + 0.3*s,
    "Ensemble_30_70": 0.3*r + 0.7*s,
    "Ensemble_Max": np.maximum(r, s)
}

truth = y.groupby(d["storm_id"]).max()
rows = []

for name, probability in models.items():

    alert = pd.Series(
        np.asarray(probability) >= 0.15
    )

    hits = alert & y.eq(1)

    storm_hits = hits.groupby(d["storm_id"]).any()
    storm_alerts = alert.groupby(d["storm_id"]).any()

    tp = int(hits.sum())
    fp = int((alert & y.eq(0)).sum())
    fn = int((~alert & y.eq(1)).sum())

    detected = int(
        storm_hits[truth.eq(1)].sum()
    )

    false_storms = int(
        storm_alerts[truth.eq(0)].sum()
    )

    rows.append({
        "Model": name,
        "Precision": precision_score(
            y, alert, zero_division=0
        ),
        "Recall": recall_score(
            y, alert, zero_division=0
        ),
        "F1": f1_score(
            y, alert, zero_division=0
        ),
        "AP": average_precision_score(
            y, probability
        ),
        "ROC_AUC": roc_auc_score(
            y, probability
        ),
        "Brier": brier_score_loss(
            y, probability
        ),
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "RI_Storms_Detected": detected,
        "RI_Storms_Missed": int(
            truth.sum() - detected
        ),
        "Non_RI_Storms_Alerted": false_storms
    })

ensemble_results = pd.DataFrame(rows)

print("\nMODEL PERFORMANCE")
display(
    ensemble_results[
        ["Model", "Precision", "Recall",
         "F1", "AP", "ROC_AUC", "Brier"]
    ].round(4)
)

print("\nDETECTION AND FALSE ALARM ANALYSIS")
display(
    ensemble_results[
        ["Model", "TP", "FP", "FN",
         "RI_Storms_Detected",
         "RI_Storms_Missed",
         "Non_RI_Storms_Alerted"]
    ]
)

print("\nEXPLORATORY RANKING BY F1")
display(
    ensemble_results.sort_values(
        "F1", ascending=False
    )[["Model", "F1", "Recall", "FP", "AP"]]
    .round(4)
)

print("\nSTEP 75F COMPLETED")
```

```python

# STEP 75G - ENSEMBLE DISAGREEMENT AUDIT

import numpy as np
import pandas as pd

print("STEP 75G - ENSEMBLE DISAGREEMENT AUDIT")
print("=" * 55)

required = [
    "storm_id", "actual_ri", "fold",
    "Regularized", "Slow_Learning"
]

assert all(c in pred.columns for c in required)

d = pred[required].copy().reset_index(drop=True)

y = d["actual_ri"].astype(int)
r = d["Regularized"].astype(float)
s = d["Slow_Learning"].astype(float)

threshold = 0.15

regular_alert = r >= threshold
slow_alert = s >= threshold
union_alert = regular_alert | slow_alert
average_alert = (0.5*r + 0.5*s) >= threshold

print("\n1. EXTRA ALERTS FROM MAX ENSEMBLE")

extra = union_alert & ~regular_alert

extra_report = d.loc[extra].copy()
extra_report["correct_alert"] = (
    extra_report["actual_ri"] == 1
)

print("Additional alerts:", int(extra.sum()))
print(
    "Additional correct alerts:",
    int((extra & y.eq(1)).sum())
)
print(
    "Additional false alarms:",
    int((extra & y.eq(0)).sum())
)

display(extra_report)

print("\n2. STORM-LEVEL RECOVERY")

truth = y.groupby(d["storm_id"]).max()

regular_hits = (
    regular_alert & y.eq(1)
).groupby(d["storm_id"]).any()

union_hits = (
    union_alert & y.eq(1)
).groupby(d["storm_id"]).any()

recovered = truth.index[
    truth.eq(1) & ~regular_hits & union_hits
]

print("Recovered RI storms:", len(recovered))
print("Storm IDs:", recovered.tolist())

for sid in recovered:
    print("\nRecovered storm:", sid)
    display(d.loc[d["storm_id"] == sid])

print("\n3. NEW FALSE ALARMS BY STORM")

false_extra = d.loc[
    extra & y.eq(0)
].groupby("storm_id").size()

display(
    false_extra.sort_values(
        ascending=False
    ).rename("extra_false_alarms").to_frame()
)

print("\n4. PERFORMANCE BY FOLD")

rows = []

for fold, g in d.groupby("fold"):
    ix = g.index
    yy = y.loc[ix]

    for name, alerts in {
        "Regularized": regular_alert,
        "Average": average_alert,
        "Max": union_alert
    }.items():

        aa = alerts.loc[ix]

        tp = int((aa & yy.eq(1)).sum())
        fp = int((aa & yy.eq(0)).sum())
        fn = int((~aa & yy.eq(1)).sum())

        precision = tp / (tp + fp) if tp + fp else 0
        recall = tp / (tp + fn) if tp + fn else 0

        rows.append({
            "Fold": fold,
            "Model": name,
            "TP": tp,
            "FP": fp,
            "FN": fn,
            "Precision": round(precision, 4),
            "Recall": round(recall, 4)
        })

fold_results = pd.DataFrame(rows)
display(fold_results)

print("\nSTEP 75G COMPLETED")
```

```python

# STEP 75H - CONFIDENCE-GATED ENSEMBLE

import numpy as np
import pandas as pd
from pathlib import Path

print("STEP 75H - CONFIDENCE-GATED ENSEMBLE")
print("=" * 55)

# Restore out-of-fold predictions
if "pred" in globals():
    df = pred.copy()
else:
    df = predictions.copy()

df = df.reset_index(drop=True)

required = [
    "storm_id", "actual_ri", "fold",
    "Regularized", "Slow_Learning"
]

assert all(c in df.columns for c in required)
assert len(df) == 1265

y = df["actual_ri"].astype(int)
r = df["Regularized"].astype(float)
s = df["Slow_Learning"].astype(float)

assert np.isfinite(r).all()
assert np.isfinite(s).all()

base_threshold = 0.15

# Storm ground truth
storm_truth = y.groupby(df["storm_id"]).max()

def evaluate(name, alerts, gate=np.nan):

    tp = int((alerts & y.eq(1)).sum())
    fp = int((alerts & y.eq(0)).sum())
    fn = int((~alerts & y.eq(1)).sum())

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    detected = (
        (alerts & y.eq(1))
        .groupby(df["storm_id"]).any()
    )

    alerted = (
        alerts.groupby(df["storm_id"]).any()
    )

    detected_storms = int(
        (storm_truth.eq(1) & detected).sum()
    )

    false_storms = int(
        (storm_truth.eq(0) & alerted).sum()
    )

    return {
        "Model": name,
        "Gate": gate,
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "RI_Storms_Detected": detected_storms,
        "Non_RI_Storms_Alerted": false_storms
    }

# Reference models
regular = r >= base_threshold
slow = s >= base_threshold
maximum = regular | slow
average = ((r + s) / 2) >= base_threshold

results = [
    evaluate("Regularized", regular),
    evaluate("Average", average),
    evaluate("Max", maximum)
]

# Gated rescue experiments
gates = [0.02, 0.03, 0.05, 0.07,
         0.09, 0.11, 0.13]

for gate in gates:

    alerts = regular | (
        slow & (r >= gate)
    )

    results.append(
        evaluate(
            f"Gated_{gate:.2f}",
            alerts,
            gate
        )
    )

summary = pd.DataFrame(results)

print("\n1. COMPLETE MODEL COMPARISON")
display(summary.round(4))

print("\n2. EXPLORATORY RANKING BY F1")
display(
    summary.sort_values(
        "F1", ascending=False
    ).round(4)
)

print("\n3. EXPLORATORY RANKING BY RECALL")
display(
    summary.sort_values(
        ["Recall", "FP"],
        ascending=[False, True]
    ).round(4)
)

print("\n4. STORMS RECOVERED BY MAX ENSEMBLE")

regular_detected = (
    (regular & y.eq(1))
    .groupby(df["storm_id"]).any()
)

max_detected = (
    (maximum & y.eq(1))
    .groupby(df["storm_id"]).any()
)

recovered = storm_truth.index[
    storm_truth.eq(1)
    & ~regular_detected
    & max_detected
]

print("Recovered storm count:", len(recovered))
print("Recovered storm IDs:", recovered.tolist())

print("\n5. EXTRA FALSE ALARMS")

extra = maximum & ~regular

print(
    "Extra correct detections:",
    int((extra & y.eq(1)).sum())
)

print(
    "Extra false alarms:",
    int((extra & y.eq(0)).sum())
)

print("\n6. SAVE EXPERIMENT")

out = Path(
    "/content/drive/MyDrive/"
    "ChakraSense_Backup/ChakraSense_V3"
)

if out.is_dir():

    file = out / "step75h_gated_ensemble.csv"
    summary.to_csv(file, index=False)

    print("Saved:", file)

else:
    print("Drive folder unavailable; results remain in memory.")

print("\nSTEP 75H COMPLETED")
```

```python

# STEP 75J - TEMPORAL CONFIRMATION EXPERIMENT
# CPU ONLY - NO GPU REQUIRED

import numpy as np
import pandas as pd
from pathlib import Path

print("STEP 75J - TEMPORAL CONFIRMATION")
print("=" * 55)

# Restore independent prediction and chronology tables
assert "predictions" in globals(), "Run Step 75B first"
assert "fold_data" in globals(), "Run Step 75B first"

p = predictions.copy().reset_index(drop=True)
f = fold_data.copy().reset_index(drop=True)

assert len(p) == len(f) == 1265

assert (
    p["storm_id"].astype(str).equals(
        f["storm_id"].astype(str)
    )
), "Storm order does not match"

assert np.array_equal(
    p["actual_ri"].astype(int).to_numpy(),
    f["ri"].astype(int).to_numpy()
), "RI labels do not match"

p["time"] = pd.to_datetime(
    f["time"], utc=True, errors="coerce"
)

assert p["time"].notna().all()
assert not p.duplicated(["storm_id", "time"]).any()

print("Chronology verified:", True)

# Existing predictions
y = p["actual_ri"].astype(bool)

p["regular_alert"] = (
    p["Regularized"].astype(float) >= 0.15
)

p["slow_alert"] = (
    p["Slow_Learning"].astype(float) >= 0.15
)

p["slow_only"] = (
    p["slow_alert"] & ~p["regular_alert"]
)

# Confirm only repeated slow-only candidates
# within the previous six hours.

p["confirmed_extra"] = False

ordered = p.sort_values(["storm_id", "time"])

for sid, group in ordered.groupby("storm_id"):
    previous = None

    for idx, row in group.iterrows():
        if not row["slow_only"]:
            continue

        t = row["time"]

        if previous is not None:
            gap = (t - previous).total_seconds() / 3600

            if 0 < gap <= 6:
                p.at[idx, "confirmed_extra"] = True

        previous = t

# Compare three methods
p["baseline_alert"] = p["regular_alert"]

p["max_alert"] = (
    p["regular_alert"] | p["slow_alert"]
)

p["temporal_alert"] = (
    p["regular_alert"] | p["confirmed_extra"]
)

def evaluate(name, column):
    a = p[column].astype(bool)

    tp = int((a & y).sum())
    fp = int((a & ~y).sum())
    fn = int((~a & y).sum())

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    positive_storms = y.groupby(p["storm_id"]).any()

    detected = (
        (a & y).groupby(p["storm_id"]).any()
    )

    alerted_storms = a.groupby(p["storm_id"]).any()

    return {
        "Method": name,
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F1": round(f1, 4),
        "RI_Storms_Detected": int(detected.sum()),
        "Non_RI_Storms_Alerted": int(
            (alerted_storms & ~positive_storms).sum()
        )
    }

results = pd.DataFrame([
    evaluate("Regularized", "baseline_alert"),
    evaluate("Max_Ensemble", "max_alert"),
    evaluate("Temporal_6h", "temporal_alert")
])

print("\n1. MODEL COMPARISON")
display(results)

# Investigate retained additional alerts
retained = p[p["confirmed_extra"]].copy()

print("\n2. CONFIRMED ADDITIONAL ALERTS")

display(
    retained[[
        "storm_id",
        "time",
        "actual_ri",
        "Regularized",
        "Slow_Learning"
    ]].sort_values(["storm_id", "time"])
)

print("\n3. TEMPORAL CONFIRMATION SUMMARY")

print("Original extra candidates:",
      int(p["slow_only"].sum()))

print("Retained extra alerts:",
      len(retained))

print("Extra correct detections:",
      int((retained["actual_ri"] == 1).sum()))

print("Extra false alerts:",
      int((retained["actual_ri"] == 0).sum()))

# Save reproducible exploratory results
out = Path(
    "/content/drive/MyDrive/"
    "ChakraSense_Backup/ChakraSense_V3"
)

if out.is_dir():
    results.to_csv(
        out / "step75j_temporal_comparison.csv",
        index=False
    )

    retained.to_csv(
        out / "step75j_confirmed_alerts.csv",
        index=False
    )

    print("\nReports saved successfully")
else:
    print("\nOutput directory unavailable")

print("\nSTEP 75J COMPLETED")
```

```python

# STEP 75K - ROBUSTNESS AND DETECTION DELAY
# CPU ONLY

import pandas as pd
import numpy as np
from pathlib import Path

print("STEP 75K - ROBUSTNESS AUDIT")
print("=" * 55)

required = [
    "storm_id", "time", "fold", "actual_ri",
    "baseline_alert", "max_alert", "temporal_alert"
]

assert "p" in globals(), "Run Step 75J first"
assert all(c in p.columns for c in required)

models = {
    "Regularized": "baseline_alert",
    "Max_Ensemble": "max_alert",
    "Temporal_6h": "temporal_alert"
}

# 1. Fold-wise performance

rows = []

for fold, df in p.groupby("fold"):
    y = df["actual_ri"].astype(bool)

    for name, col in models.items():
        a = df[col].astype(bool)

        tp = int((a & y).sum())
        fp = int((a & ~y).sum())
        fn = int((~a & y).sum())

        precision = tp / (tp + fp) if tp + fp else 0
        recall = tp / (tp + fn) if tp + fn else 0

        f1 = (
            2 * precision * recall / (precision + recall)
            if precision + recall else 0
        )

        rows.append({
            "Fold": fold,
            "Model": name,
            "TP": tp,
            "FP": fp,
            "FN": fn,
            "Precision": round(precision, 4),
            "Recall": round(recall, 4),
            "F1": round(f1, 4)
        })

fold_report = pd.DataFrame(rows)

print("\n1. FOLD-WISE PERFORMANCE")
display(fold_report)

# 2. Storm-level detection comparison

storm_rows = []

for sid, df in p.groupby("storm_id"):
    actual = df["actual_ri"].astype(bool)

    if not actual.any():
        continue

    row = {
        "storm_id": sid,
        "RI_windows": int(actual.sum())
    }

    for name, col in models.items():
        hits = df[col].astype(bool) & actual
        row[name + "_hits"] = int(hits.sum())

        if hits.any():
            first_hit = df.loc[hits, "time"].min()
            first_ri = df.loc[actual, "time"].min()

            delay = (
                first_hit - first_ri
            ).total_seconds() / 3600

            row[name + "_delay_h"] = delay
        else:
            row[name + "_delay_h"] = np.nan

    storm_rows.append(row)

storm_report = pd.DataFrame(storm_rows)

print("\n2. RI STORM DETECTION COMPARISON")
display(storm_report)

# 3. Determine which additional alerts survived

extra = p[
    p["temporal_alert"] &
    ~p["baseline_alert"]
].copy()

print("\n3. ADDITIONAL TEMPORAL ALERTS")
display(extra[[
    "storm_id", "time", "fold",
    "actual_ri",
    "Regularized",
    "Slow_Learning"
]].sort_values(["storm_id", "time"]))

# 4. Check temporal confirmation delays

print("\n4. CONFIRMATION DELAY SUMMARY")

for name in models:
    col = name + "_delay_h"
    valid = storm_report[col].dropna()

    print(name)
    print("Detected RI storms:", len(valid))
    print("Median first-hit delay (hours):",
          round(valid.median(), 2) if len(valid) else None)

# 5. Save reports

out = Path(
    "/content/drive/MyDrive/"
    "ChakraSense_Backup/ChakraSense_V3"
)

if out.is_dir():
    fold_report.to_csv(
        out / "step75k_fold_robustness.csv",
        index=False
    )

    storm_report.to_csv(
        out / "step75k_storm_detection.csv",
        index=False
    )

    print("\nReports saved successfully")

print("\nSTEP 75K COMPLETED")
```

```python

# STEP 75M.3 - FINAL TRAINING READINESS AUDIT
# CPU ONLY - NO MODEL TRAINING

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.base import BaseEstimator
from sklearn.utils.validation import check_is_fitted

root = Path("/content/drive/MyDrive/ChakraSense_Backup")
v3 = root / "ChakraSense_V3"

X = pd.read_csv(v3 / "training_features_v3.csv")

print("STEP 75M.3 - TRAINING READINESS")
print("=" * 55)

print("\n1. TRAINING FEATURES")
print("Shape:", X.shape)
print("Feature names:", X.columns.tolist())

print("\nMissing values by feature:")
print(X.isna().sum()[X.isna().sum() > 0])

print("\n2. FIND TRAINING LABELS")

candidates = []

for name in ["fold_data", "train", "data"]:
    obj = globals().get(name)

    if isinstance(obj, pd.DataFrame) and "ri" in obj:
        print(
            name,
            "| rows:", len(obj),
            "| RI cases:", int(obj["ri"].sum())
        )

        if len(obj) == len(X):
            candidates.append((name, obj))

print("\n3. FEATURE ALIGNMENT AUDIT")

for name, df in candidates:
    matched = []
    mismatched = []

    for col in X.columns:
        for candidate in [col, "v3_" + col]:
            if candidate not in df.columns:
                continue

            a = pd.to_numeric(X[col], errors="coerce").to_numpy()
            b = pd.to_numeric(
                df[candidate], errors="coerce"
            ).to_numpy()

            agreement = np.isclose(
                a, b,
                rtol=1e-6,
                atol=1e-6,
                equal_nan=True
            )

            if agreement.all():
                matched.append(col)
            else:
                mismatched.append(
                    (col, int((~agreement).sum()))
                )

            break

    print("\nCandidate:", name)
    print("Matching feature columns:", len(matched))
    print("Examples:", matched[:10])
    print("Mismatches:", mismatched[:10])

print("\n4. CLASSIFIER CONFIGURATION")

obj = globals().get("clf")

if isinstance(obj, BaseEstimator):
    print("Model:", type(obj).__name__)

    print("\nParameters:")
    for key, value in obj.get_params().items():
        print(key, ":", value)

    try:
        check_is_fitted(obj)
        print("\nClassifier fitted: True")
    except Exception:
        print("\nClassifier fitted: False")
else:
    print("No classifier named clf found")

print("\n5. AVAILABLE MODEL CONFIGURATIONS")

names = [
    "models", "model_configs", "configs",
    "param_grid", "model_params", "variants",
    "regularized_model", "slow_learning_model"
]

for name in names:
    if name in globals():
        value = globals()[name]
        print(name, ":", str(value)[:700])

print("\nSTEP 75M.3 COMPLETED")
print("No models modified or trained.")
print("Independent test remains untouched.")
```

```python
from pathlib import Path
import io,json,zipfile,os
os.environ['OMP_NUM_THREADS']='2'
import pandas as pd,numpy as np,joblib,sklearn
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import confusion_matrix
from functools import lru_cache
from google.colab import drive
drive.mount('/content/drive')
root = Path('/content/drive/MyDrive/ChakraSense_Backup')
def blob(n):
 return root / n
def csv(n):return pd.read_csv(blob(n))
v='ChakraSense_V3/'
data=csv('Combined_Real_Dataset/combined_samples_with_splits.csv')
def split(name):return data[data.split==name].reset_index(drop=True)
train=split('train')
X=csv(v+'training_features_v3.csv')
X['current_wind']=train.wind_kt
o=csv(v+'v3_ri_oof_predictions.csv')
assert (o.storm_id==train.storm_id).all() and (o.actual_ri==train.ri).all()
configs={'Regularized':(.05,60,5,40,10),'Slow_Learning':(.035,140,5,25,5)}
models={}
for name,(rate,it,leaf,minimum,l2) in configs.items():
 m=HistGradientBoostingClassifier(learning_rate=rate,max_iter=it,max_leaf_nodes=leaf,min_samples_leaf=minimum,l2_regularization=l2,early_stopping=False,random_state=42)
 for f in o.fold.unique():
  val=o.fold==f
  m.fit(X[~val],train.ri[~val])
  p=m.predict_proba(X[val])[:,1]
  print(name,f,'max probability difference',max(abs(p-o.loc[val,name])),flush=True)
  assert np.allclose(p,o.loc[val,name],atol=1e-7)
 models[name]=m.fit(X,train.ri)
wind=csv(v+'original_wind_records.csv')
wind['time']=pd.to_datetime(wind.time,utc=True)
@lru_cache(maxsize=1)
def frames(path):
 with np.load(blob(path.split('ChakraSense_Backup/')[1]),allow_pickle=False) as d:return d['frames']
def features(rows):
 out=[]
 for r in rows.itertuples():
  ids=json.loads(r.frame_indices)
  x=frames(r.file_path)[[ids[0],ids[-1]]]
  x=((x-250.)/50.).astype('float32')*50+250
  a={'current_wind':r.wind_kt}
  w=wind[wind.storm_id==r.storm_id].sort_values('time').drop_duplicates('time',keep='last')
  for h in [0,6,12,24]:
   t=pd.Timestamp(r.time)-pd.Timedelta(hours=h)
   prev=w[w.time<=t].tail(1)
   age=(t-prev.time.iloc[0]).total_seconds()/3600 if len(prev) else np.nan
   speed=prev.wind_kt.iloc[0] if age<=6 else np.nan
   if h==0: now=speed
   else:a[f'wind_change_{h}h']=now-speed
   a[f'wind_age_{h}h' if h else 'current_wind_age']=age if age<=6 else np.nan
  for j,ch in enumerate(['IR','WV']):
   first,last=x[:,j]
   for k,value in [('mean',last.mean()),('std',last.std()),('min',last.min()),('centre_mean',last[40:88,40:88].mean()),('change',last.mean()-first.mean()),('p10',np.percentile(last,10))]:a[ch+'_'+k]=value
  a['IR_cold_cloud_fraction']=(x[-1,0]<230).mean()
  out.append(a)
 return pd.DataFrame(out)[X.columns]
dev=split('development')
ref=csv(v+'development_features_v3.csv')
ref['current_wind']=dev.wind_kt
assert np.allclose(features(dev),ref,equal_nan=True,atol=1e-4)
out = root / 'ChakraSense_V3' / 'final_run'
out.mkdir(exist_ok=True)
path=out/'ChakraSense_V3.joblib'
joblib.dump(dict(models=models,features=list(X),threshold=.15,confirmation_hours=6,sklearn_version=sklearn.__version__),path)
bundle=joblib.load(path)
test=split('test')
assert set(test.storm_id).isdisjoint(train.storm_id) and set(test.storm_id).isdisjoint(dev.storm_id)
T=features(test)
results=test[['sample_id','storm_id','time','ri']].copy()
for name,m in bundle['models'].items():results[name]=m.predict_proba(T)[:,1]
regular=results.Regularized>=.15
slowonly=(results.Slow_Learning>=.15)&~regular
times=pd.to_datetime(results.loc[slowonly,'time'],utc=True).sort_values()
gaps=times.groupby(results.storm_id).diff().dt.total_seconds()/3600
results['alert']=regular | (gaps.gt(0)&gaps.le(6)).reindex(results.index,fill_value=False)
results.to_csv(out/'independent_predictions.csv',index=False)
print('test confusion matrix',confusion_matrix(results.ri,results.alert))
```
